## Rank Distrubution
Here we are checking the skill distrubution of our player sample to ensure the methodoligy of sampling is correct.
We know that the skill distribution should be normal bell curve.

The tiers in Team Fight Tactics are (from worst to best):
- Iron
- Bronze
- Silver
- Gold
- Platnium
- Emerald
- Diamond
- Master
- Grandmaster
- Challenger

The apex tiers are extremely small, probably <3% of total population, so a historgram of population count will look right skewed, but it's not.
With the below chart, we can see that our sample is represenative of the population and our data gathering method is sufficient and acceptable for analysis. 

```python
tier_order = [
    'IRON', 'BRONZE', 'SILVER', 'GOLD',
    'PLATINUM', 'EMERALD', 'DIAMOND', 'MASTER', 
    'GRANDMASTER', 'CHALLENGER'
]

# Representative colors for each tier
tier_colors = {
    'IRON': '#484848',          # Dark gray
    'BRONZE': '#cd7f32',        # Bronze
    'SILVER': '#c0c0c0',        # Silver
    'GOLD': '#ffd700',          # Gold
    'PLATINUM': '#00bfff',      # Deep Sky Blue
    'EMERALD': '#50c878',       # Emerald Green
    'DIAMOND': '#b9f2ff',       # Diamond Light Blue
    'MASTER': '#aa00ff',        # Purple
    'GRANDMASTER': '#ff4500',   # Orange Red
    'CHALLENGER': '#1e90ff',    # Dodger Blue
}

# Count and reindex to enforce tier order
tier_counts = final_df['rank_at_match_time.tier'].value_counts()
tier_counts = tier_counts.reindex(tier_order, fill_value=0)

# Convert categorical tiers to numeric x values
x = np.arange(len(tier_order))
y = tier_counts.values

# Smooth curve with spline
spline_x = np.linspace(x.min(), x.max(), 300)
spline = make_interp_spline(x, y, k=3)
spline_y = spline(spline_x)

# Get bar colors in order
bar_colors = [tier_colors[tier] for tier in tier_order]

# Plot
plt.figure(figsize=(10, 6))
plt.bar(x, y, tick_label=tier_order, color=bar_colors)
plt.plot(spline_x, spline_y, color='black', linewidth=2, label='Spline Curve')

plt.title('Distribution of Ranks at Match Time')
plt.xlabel('Tier')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

os.makedirs("html", exist_ok=True)
fig.write_html("html/rank_distribution.html", include_plotlyjs="cdn")
plt.show()
```

# WinRates by Traits Levels

```python

df = tftinsights.data.get_tft_dataset()
df_high_rank = tftinsights.data.get_tft_dataset(rankFilter = ['DIAMOND','MASTER','GRANDMASTER','CHALLENGER'])

#df[df['rank_at_match_time.tier'].isin (['DIAMOND','MASTER','GRANDMASTER','CHALLENGER'])]

def transform_trait_data ( df ):
    # Extract trait columns
    traits_level_data = pd.read_csv('csv/traits_csv.csv')
    
    trait_columns = [col for col in df.columns if col.startswith('trait_')]
    
    #print ( trait_columns )
    
    # Melt the dataframe to long format so we have levels of traits 
    trait_df = df.melt(id_vars='placement', value_vars=trait_columns,
                       var_name='trait', value_name='trait_level')

    trait_df['trait'] = trait_df['trait'].str.replace('trait_', '')
    # Filter out zero-level traits (not active)
    trait_df = trait_df[trait_df['trait_level'] > 0]
    
    # Group by trait and level
    grouped = trait_df.groupby(['trait', 'trait_level'])
    
    # Compute statistics
    trait_stats = grouped.agg(
        frequency=('placement', 'count'),
        avg_placement=('placement', 'mean'),
        top4_rate=('placement', lambda x: (x <= 4).mean()),
        top1_rate=('placement', lambda x: (x == 1).mean())
    ).reset_index()
    

    
    trait_stats['avg_placement'] = trait_stats['avg_placement'].round(1)
    trait_stats['trait_level'] = trait_stats['trait_level'].astype(int)
    #trait_stats['top4_rate'] = (trait_stats['top4_rate'] * 100).round(1)
    #trait_stats['top1_rate'] = (trait_stats['top1_rate'] * 100).round(1)
    trait_stats.rename(columns={
        'top4_rate': 'top4_rate_pct',
        'top1_rate': 'top1_rate_pct'
    }, inplace=True)
    
    pd.set_option('display.max_rows', 100)
    
    trait_stats.sort_values(by=['top4_rate_pct' ], ascending=False, inplace=True)
    trait_stats.reset_index(inplace=True)

    trait_stats = trait_stats.merge(
    traits_level_data,
    left_on=['trait', 'trait_level'],  # Columns from trait_stats
    right_on=['id', 'tier_current'],    # Columns from traits_level_data
    how='left'
)
    
    trait_stats.to_csv( 'csv/trait_stats.csv')

    
    trait_stats['trait'] = trait_stats.apply(
        lambda row: f"{row['trait']}_{row['trait_level']}", axis=1
    )

    return trait_stats

trait_stats =  transform_trait_data ( df )
high_rank_trait_stats =transform_trait_data ( df_high_rank )
```

```python
import plotly.graph_objects as go
def top4_top1_trait_win_rate ( trait_stats , name=1):
    
    # Manually scale sizes if desired (optional multiplier)
    scaled_size = trait_stats['num_units'] * 2
    
    # Format hover text
    hover_text = [
        f"<b>{row['name']}</b><br>"
        f"Trait: {row['trait']}<br>"
        f"Num Units: {row['num_units']}<br>"
        f"Top 4 Rate: {row['top4_rate_pct']:.4%}<br>"
        f"Top 1 Rate: {row['top1_rate_pct']:.4%}<br>"
        f"Trait Level: {row['trait_level']}<br>"
        f"Tier Current: {row['tier_current']}<br>"
        f"Count: {row['frequency']}"
        for _, row in trait_stats.iterrows()
    ]
    
    fig = go.Figure()
    
    fig.add_trace(go.Scatter(
        x=trait_stats['top4_rate_pct'],  # e.g. 0.35 for 35%
        y=trait_stats['top1_rate_pct'],  # e.g. 0.10 for 10%
        mode='markers',
        marker=dict(
            size=scaled_size,
            sizemode='diameter',
            sizemin=4,
            opacity=0.7,
            color=trait_stats['num_units'],
            colorscale='Viridis',
            colorbar=dict(title='num_units')
        ),
        text=hover_text,
        hoverinfo='text',
    ))
    
    # Update layout
    fig.update_layout(
        title="TFT Traits: Top 4 Rate vs Top 1 Rate",
        xaxis_title="Top 4 Rate (%)",
        yaxis_title="Top 1 Rate (%)",
        width=800,
        height=600
    )

    # Horizontal line at 20% (y = 0.2)
    fig.add_shape(
        type="line",
        x0=-0.1, x1=2,  # full width of the x-axis
        y0=0.2, y1=0.2,
        line=dict(color="gray", width=2, dash="dash"),
        xref="x", yref="y"
    )

    # Vertical line at 50% (x = 0.5)
    fig.add_shape(
        type="line",
        x0=0.5, x1=0.5,
        y0=-0.1, y1=2,  # full height of the y-axis
        line=dict(color="gray", width=2, dash="dash"),
        xref="x", yref="y"
    )
    
    # Optional: convert percentages to axis ticks
    fig.update_xaxes(tickformat=".0%")
    fig.update_yaxes(tickformat=".0%")

    # Optional: convert percentages to axis ticks and fix axis range
    fig.update_xaxes(tickformat=".0%", range=[-0.1, 1.1])
    fig.update_yaxes(tickformat=".0%", range=[-0.1, 1.1])

    fig.write_html(f"html/trait_win_rate_{name}.html", include_plotlyjs="cdn")
    fig.show()
```

# Win Rate by Units

```python
from tftinsights.units import get_unit_list
unit_list = get_unit_list (units_only=True, add_prefix=True)
df = tftinsights.data.get_tft_dataset() 
unit_win_rate_df = pd.DataFrame(columns=['unit', 'level_1', 'level_2', 'level_3', 'level_4'])

for unit_key in unit_list:
    if unit_key in df.keys():
        level_1_rows = df[df[unit_key] == 1]
        level_2_rows = df[df[unit_key] == 2]
        level_3_rows = df[df[unit_key] == 3]
        level_4_rows = df[df[unit_key] == 4]
        
        level_1_winrate = level_1_rows['win'].mean() if not level_1_rows.empty else 0
        level_2_winrate = level_2_rows['win'].mean() if not level_2_rows.empty else 0
        level_3_winrate = level_3_rows['win'].mean() if not level_3_rows.empty else 0
        level_4_winrate = level_4_rows['win'].mean() if not level_4_rows.empty else 0
            
        unit_win_rate_df.loc[len(unit_win_rate_df)] = [unit_key, level_1_winrate, level_2_winrate, level_3_winrate, level_4_winrate]
```

```python
import plotly.graph_objects as go

levels = ['level_1', 'level_2', 'level_3', 'level_4']
x_all = [1, 2, 3, 4]

fig = go.Figure()

for _, row in unit_win_rate_df.iterrows():
    unit_name = row['unit'].replace('unit_TFT14_', '')  # clean unit name
    y_vals = row[levels].values.astype(float)
    
    if y_vals[-1] == 0:
        x = x_all[:-1]
        y = y_vals[:-1]
    else:
        x = x_all
        y = y_vals
    
    fig.add_trace(go.Scatter(
        x=x,
        y=y,
        mode='lines+markers',
        name=unit_name,
        hovertemplate=(
            f"<b>{unit_name}</b><br>" +
            "Level %{x}<br>" +
            "Win Rate: %{y:.2f}<extra></extra>"
        )
    ))

fig.update_layout(
    title='Win Rate by Unit and Star Level (Skipping Level 4 if 0)',
    xaxis_title='Star Level',
    yaxis_title='Win Rate',
    xaxis=dict(tickmode='array', tickvals=x_all),
    yaxis=dict(range=[0, 1]),
    #legend_title_text='Units',
    showlegend=False,
    width=900,
    height=600
)

fig.show()
```

# Items

We'll skip items for now because it's complicated


```python
import pandas as pd
import re

# Identify all item columns
player_df = tftinsights.data.get_tft_dataset(include_items=False,include_traits=False,include_units_item=True) 
item_cols = [col for col in player_df.columns if re.match(r'.*_item_[0-2]$', col)]

# Group columns by unit name
unit_to_items = {}
for col in item_cols:
    match = re.match(r'(.*?)_item_\d$', col)
    if match:
        unit = match.group(1)
        unit_to_items.setdefault(unit, []).append(col)
        

# Collect results
rows = []

for unit, cols in unit_to_items.items():
    # Only keep rows where the unit appears (at least one item is non-NaN and non-zero)
    unit_df = player_df[cols]
    unit_rows = unit_df.notna().any(axis=1)
    relevant_rows = unit_df[unit_rows]

    if relevant_rows.empty:
        continue

    # Count valid items (exclude 0s and NaNs)
    item_counts = (relevant_rows.map(lambda x: x != 0 and pd.notna(x))).sum(axis=1)

    # Compute probabilities
    dist = item_counts.value_counts().reindex([0, 1, 2, 3], fill_value=0)
    total = dist.sum()
    probs = (dist / total).to_dict()

    rows.append({
        'unit': unit,
        'item_0': probs.get(0, 0.0),
        'item_1': probs.get(1, 0.0),
        'item_2': probs.get(2, 0.0),
        'item_3': probs.get(3, 0.0),
    })

# Final DataFrame
unit_item_distribution_df = pd.DataFrame(rows)
```

```python
import pandas as pd
from collections import Counter
import re

# Extract all item columns
item_cols = [col for col in player_df.columns if re.match(r'^TFT14_.*_item_[0-2]$', col)]

# Group item columns by unit
unit_to_items = {}
for col in item_cols:
    match = re.match(r'^TFT14_(.*?)_item_\d$', col)
    if match:
        unit = match.group(1)
        unit_to_items.setdefault(unit, []).append(col)

# Compute most common item(s) for each unit
most_common_items = {}

for unit, cols in unit_to_items.items():
    all_items = player_df[cols].values.flatten()
    all_items = all_items[~pd.isna(all_items)]  # remove NaNs
    all_items = [item for item in all_items if item != 0]  # remove item 0
    counter = Counter(all_items)
    most_common = counter.most_common()
    most_common_items[unit] = most_common

# Flatten the dictionary into a list of rows
rows = []
for unit, items in most_common_items.items():
    for item, count in items:
        rows.append({'unit': f'TFT14_{unit}', 'item': item, 'count': count})

# Create the DataFrame
most_common_items_df = pd.DataFrame(rows)

# Reorder columns: item, unit, count
most_common_items_df_sorted = most_common_items_df.sort_values(
    by=['item', 'count'], ascending=[True, False]
)[['item', 'unit', 'count']].reset_index(drop=True)


```

```python
filtered_items_df = most_common_items_df_sorted[
    ~most_common_items_df_sorted['item'].str.contains('Artifact|Emblem|Radiant', case=False, na=False)
].copy()

# Sort by item and descending count
sorted_basic_items_df = filtered_items_df.sort_values(['item', 'count'], ascending=[True, False])

# Get top 8 units for each Artifact item
top_5_item_users = sorted_basic_items_df.groupby('item', group_keys=False).head(8).copy()

# Add rank to each unit per item based on count
top_5_item_users['rank'] = (
    top_5_item_users.groupby('item')['count']
    .rank(method='first', ascending=False)
    .astype(int)
)

# Pivot to wide format for unit and count
unit_pivot = top_5_item_users.pivot(index='item', columns='rank', values='unit')
count_pivot = top_5_item_users.pivot(index='item', columns='rank', values='count')

# Rename columns for interleaving
unit_pivot.columns = [f"unit_{i}" for i in unit_pivot.columns]
count_pivot.columns = [f"count_{i}" for i in count_pivot.columns]

# Concatenate and reorder to interleave unit and count
wide_item_df = pd.concat([unit_pivot, count_pivot], axis=1)

# Reorder columns: unit_1, count_1, unit_2, count_2, ...
cols = []
for i in range(1, 6):  # Top 5 units (adjustable)
    if f"unit_{i}" in wide_item_df.columns and f"count_{i}" in wide_item_df.columns:
        cols.extend([f"unit_{i}", f"count_{i}"])

wide_item_df = wide_item_df[cols]
wide_item_df.insert(0, 'item', unit_pivot.index)  # Add item back as first column

# Sort by count_1 descending
#wide_item_df_sorted = wide_item_df.sort_values(by='count_1', ascending=False).reset_index(drop=True)

# Compute row-wise sum of all counts
count_cols = [col for col in wide_item_df.columns if col.startswith('count_')]
wide_item_df['total_count'] = wide_item_df[count_cols].sum(axis=1)

# Compute proportion of count_1 to total
wide_item_df['count_1_proportion'] = wide_item_df['count_1'] / wide_item_df['total_count']

# Sort by this proportion descending
wide_item_df_sorted = wide_item_df.sort_values(by='count_1_proportion', ascending=False).reset_index(drop=True)

# Display configuration
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_rows', 200)

# Display result
wide_item_df_sorted
```

```python
def graph_item_top_5 (df , filterString , columns=7) :

    import plotly.graph_objects as go
    from plotly.subplots import make_subplots
    import re
    
    # Parameters
    num_items = len(df)
    cols = columns
    rows = -(-num_items // cols)  # Ceiling division
    
    # Define consistent colors for top 1–5 positions
    fixed_colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']  # blue, orange, green, red, purple
    
    # Extract trait names for titles
    def extract_name(item):
        print (item)
        match = re.search(filterString, item)
        return match.group(1) if match else item
    
    titles = df['item'].apply(extract_name)
    
    # Create subplot grid
    fig = make_subplots(
        rows=rows, cols=cols,
        specs=[[{'type': 'domain'}] * cols] * rows,
        subplot_titles=titles.tolist()
    )
    
    # Plot each pie chart
    for idx, row in df.iterrows():
        r = idx // cols + 1
        c = idx % cols + 1
        labels = [row[f'unit_{i}'] for i in range(1, 6)]
        values = [row[f'count_{i}'] for i in range(1, 6)]
        max_index = values.index(max(values))
        label_texts = [label if i == max_index else '' for i, label in enumerate(labels)]
    
        fig.add_trace(
            go.Pie(
                labels=labels,
                values=values,
                text=label_texts,
                textinfo='text',
                hole=0.3,
                marker=dict(colors=fixed_colors, line=dict(color='black', width=1)),
                textfont=dict(size=10)
            ),
            row=r, col=c
        )
    
    # Layout customization
    fig.update_layout(
        height=rows * 220,
        width=cols * 220,
        title_text="Top 5 Unit Distribution per Item",
        title_font_size=18,
        showlegend=False,
        margin=dict(t=50, b=20, l=20, r=20),
        font=dict(size=10)
    )
    
    fig.show()
```
